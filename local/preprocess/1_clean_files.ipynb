strokes number pattern looks like this

```
<g id="kvg:StrokeNumbers_0f9a8" style="...">
	<text transform="matrix(1 0 0 1 42.50 15.13)">1</text>
		...
	<text transform="matrix(1 0 0 1 36.4 65.5)">5</text>
</g>
```


- find 
- starting with ```<g id="kvg:StrokeNumbers```
- ending with ```</g>```
- replace with ""

xml information pattern looks like this

```
<?xml version="1.0" encoding="UTF-8"?>
  ...
]>
```

- find 
- starting with ```<?xml version="1.0" encoding="UTF-8"?>```
- ending with ```]>```
- replace with ""


````python
import re

pattern_stroke_number = r'<g id="kvg:StrokeNumbers_?[^\"]*".*?</g>'
pattern_kvg_mark = r'kvg:'
pattern_xml_info = r'<\?xml version="1.0" encoding="UTF-8"\?>?[^\"]*".*?]>'

# Replacing the matched pattern with an empty string
removed_xml_info = re.sub(pattern_xml_info, '', example_text, flags=re.DOTALL)
removed_stroke_numbers = re.sub(pattern_stroke_number, '', removed_xml_info, flags=re.DOTALL)
result = re.sub(pattern_kvg_mark, '', removed_stroke_numbers, flags=re.DOTALL).strip()

print("Removing pattern from example text...")
print("Expected:", final)
print("Result:", result)
print("Worked:", result == final)
````

````python
# paths
raw_kanji_main_path = "../data_raw/kanji_main"
raw_kanji_all_path = "../data_raw/kanji_all"

# save to
cleaned_kanji_main_path = "../data_cleaned/kanji_main"
cleaned_kanji_all_path = "../data_cleaned/kanji_all"
````

````python
import os

for path in [cleaned_kanji_main_path, cleaned_kanji_all_path]:
  # Check if the folder exists
  if not os.path.exists(path):
      # Create the folder
      os.makedirs(path)
      print(f"Folder '{path}' created.")
  else:
      print(f"Folder '{path}' already exists.")
````

````python
import sys
import shutil

def clean_kanji_svgs(raw_path, save_path):
  # delete everything from save_path
  shutil.rmtree(save_path)
  os.makedirs(save_path)

  file_names = os.listdir(raw_path)

  total = len(file_names)
  processed = 0

  for file_name in file_names:
    processed += 1

    sys.stdout.write(f"\rProcessing... {int((processed / total) * 100)}%, remaining: {processed}/{total}")
    sys.stdout.flush()
    file_path = os.path.join(raw_path, file_name)

    with open(file_path, 'r') as file:
      content = file.read()
      
      # removing xml info
      removed_xml_info = re.sub(pattern_xml_info, '', content, flags=re.DOTALL)
      # removing stroke number
      removed_stroke_numbers = re.sub(pattern_stroke_number, '', removed_xml_info, flags=re.DOTALL)
      # removing kvg: tag
      result = re.sub(pattern_kvg_mark, '', removed_stroke_numbers, flags=re.DOTALL).strip()
      file.close()

    new_file_path = os.path.join(save_path, file_name)
    
    with open(new_file_path, 'w') as new_file:
      new_file.write(result)
      new_file.close()
````

````python
# process kanji all svg
clean_kanji_svgs(raw_kanji_all_path, cleaned_kanji_all_path)
````

````python
# process kanji main svg
clean_kanji_svgs(raw_kanji_main_path, cleaned_kanji_main_path)
````
