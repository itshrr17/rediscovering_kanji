So far, in our data, numbers, alphabets, hiragana, and katakana are included.
- We don't need this data,
- we will apply clustering on kanji only
- and only on jouyou kanji - 2136 kanji

Since, our objective is to find patterns in used kanji to help learning japanese kanji

```python
import pandas as pd

jouyou_kanji_list_file_path = "../data_raw/jouyou_kanji_list.csv"

jouyou_kanji_list = pd.read_csv(jouyou_kanji_list_file_path)

print(jouyou_kanji_list)
```

```python
jouyou_kanji_list.shape
```

```python
jouyou_look_up = set(jouyou_kanji_list["kanji"])
```

```python
import os
import shutil

jouyou_folder_path = "../data_cleaned/jouyou_kanji"

# create folder for it, it already exists delete and create new
if os.path.exists(jouyou_folder_path):
  shutil.rmtree(jouyou_folder_path)

os.makedirs(jouyou_folder_path)
```

```python
cleaned_kanji_main_path = "../data_cleaned/kanji_main"
cleaned_kanji_all_path = "../data_cleaned/kanji_all" # not using it
```

```python
import json
filename_to_kanji_map_file_path = "../data_raw/svg_to_kanji_index.json"

with open(filename_to_kanji_map_file_path) as f:
  filename_to_kanji_mapping = json.load(f)

filename_to_kanji_mapping
```

```python
import sys


file_names = os.listdir(cleaned_kanji_main_path)

total = len(file_names)
processed = 0
copied = 0

for file_name in file_names:
    processed += 1

    sys.stdout.write(f"\rProcessing... {int((processed / total) * 100)}%; Copied: {copied}; remaining: {processed}/{total}")
    sys.stdout.flush()

    if not file_name in filename_to_kanji_mapping: # skip if file_name is not in mapping
        continue

    kanji_char = filename_to_kanji_mapping[file_name]

    if not kanji_char in jouyou_look_up: # skip if not in jouyou_kanji_list
        continue

    file_path = os.path.join(cleaned_kanji_main_path, file_name)
    new_path = os.path.join(jouyou_folder_path, file_name)
    copied += 1

    shutil.copyfile(src=file_path, dst=new_path)
```
