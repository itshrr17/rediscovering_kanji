```python
# paths
jouyou_kanji_raster_folder_path = "../data_cleaned/jouyou_kanji_raster"
all_kanji_raster_folder_path = "../data_cleaned/all_kanji_raster"

paths = [jouyou_kanji_raster_folder_path, all_kanji_raster_folder_path]
```

```python
import os
import shutil

jouyou_kanji_grayscale_folder_path = "../data_cleaned/jouyou_kanji_grayscale"
all_kanji_grayscale_folder_path = "../data_cleaned/all_kanji_grayscale"

dest_paths = [jouyou_kanji_grayscale_folder_path, all_kanji_grayscale_folder_path]

for path in dest_paths:
  # Check if the folder exists
    if os.path.exists(path):
        shutil.rmtree(path)
        os.makedirs(path)
    else:
        os.makedirs(path)

    print(f"Folder '{path}' created.")
```

```python
import cv2
import numpy as np

# converts the image from png to B&W image
def grayscale(image_path):
    img = cv2.imread(image_path, cv2.IMREAD_UNCHANGED)
    if img is None:
        raise ValueError(f"Failed to load image: {image_path}")
    if img.shape[2] == 4:
        alpha = img[:, :, 3]
        gray = cv2.cvtColor(img[:, :, :3], cv2.COLOR_BGR2GRAY)
        mask = (gray < 128) & (alpha > 0)
        binary = np.zeros_like(gray, dtype=np.uint8)
        binary[mask] = 255
    else:
        gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
        _, binary = cv2.threshold(gray, 128, 255, cv2.THRESH_BINARY_INV)
    return binary
```

```python
from glob import glob
from tqdm import tqdm

# conversion
source_paths = ['../data_cleaned/all_kanji_raster/*', '../data_cleaned/jouyou_kanji_raster/*']

for source_path, dest_path in zip(source_paths, dest_paths):
    os.makedirs(dest_path, exist_ok=True)
    image_paths = glob(source_path)
    
    for path in tqdm(image_paths):
        binary = grayscale(path)
        filename = os.path.basename(path)
        save_path = os.path.join(dest_path, filename)
        cv2.imwrite(save_path, binary)
```
