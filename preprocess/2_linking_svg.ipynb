# Linking SVG to kanji (utf-8)

#### Since we already have kanji to SVG mapping
in ```../data_raw/kanji_to_svg_index.json```

- ```some_kanji_character``` -> ```svg_file_name.svg```

#### We need to reverse the mapping so we could find the kanji from SVG file naming
- ```svg_file_name.svg``` -> ```some_kanji_character```

#### We will have two way connection
- ```some_kanji_character``` <-> ```svg_file_name.svg```


##### From ```some_kanji_character``` we can find kanji information

````python
import json

kanji_svg_mapping_path = "../data_raw/kanji_to_svg_index.json"

with open(kanji_svg_mapping_path) as f:
  mapping = json.load(f)
  f.close()

new_mapping = {}

for k, v in mapping.items():
  for s in v:
    new_mapping[s] = k

svg_to_kanji_mapping_path = "../data_raw/svg_to_kanji_index.json"

with open(svg_to_kanji_mapping_path, 'w+') as sf:
  dump = json.dumps(obj= new_mapping, indent=2, ensure_ascii=False) # ensure_ascii=False, will prevent kanji (non-ascii) to be converted to unicode escape
  sf.write(dump)
  sf.close()
````

````python
with open(svg_to_kanji_mapping_path, 'r') as f:
  print(f.read())
  f.close()
````

Now we have a way to find kanji from svg
- next is finding kanji information from kanji

### linking
```svg -> kanji -> kanji_info```

````python
jouyou_kanji_info_file_path = "../data_raw/jouyou_kanji_information_all.csv"

import pandas as pd

df = pd.read_csv(filepath_or_buffer=jouyou_kanji_info_file_path, delimiter=';')
df.head()
````

````python
def get_kanji_info(kanji_char):
  return df[df["Kanji"] == kanji_char]
````
